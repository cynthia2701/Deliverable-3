```python
# Character specialization

from character import Character, Warrior, Mage
from dice import Dice
from items import Weapon

# Warrior:
link = Warrior("Link", hp_max=120, damage=30, armor=5)
print(link)

print(f"{link.name} HP before damage: {link.hp}")
link.take_damage(20)
print(f"{link.name} HP after damage: {link.hp}")


# Mage:
zelda = Mage("Zelda", hp_max=100, damage=15)
moblin = Character("Moblin", hp_max=100, damage=10)

print(f"\n{zelda.name} mana before attack: {zelda.mana}")
print(f"{moblin.name} HP before attack: {moblin.hp}")

zelda.attack(moblin)

print(f"{zelda.name} mana after attack: {zelda.mana}")
print(f"{moblin.name} HP after attack: {moblin.hp}")


# Mage with weapon and low mana:
staff = Weapon("Magic Staff", damage_dice=Dice(6))
zelda_weapon = Mage("Zelda", hp_max=100, damage=15,
                    weapon=staff, mana=10)
moblin2 = Character("Moblin", hp_max=100)

# Magic attack does not use the weapon.
zelda_weapon.attack(moblin2)
print(f"\nMoblin HP after magic attack: {moblin2.hp}")
print(f"Zelda mana after magic attack: {zelda_weapon.mana}")

# Normal attack uses the weapon when mana is below 10.
zelda_weapon.attack(moblin2)
print(f"Moblin HP after weapon attack: {moblin2.hp}")
print(f"Zelda mana after weapon attack: {zelda_weapon.mana}")
```

 Warrior and Mage classes, which inherit from Character. The Warrior uses armor to reduce incoming damage, while the Mage uses mana to perform magic attacks that deal double damage. The examples show how these abilities work.

```python
# Battle simulation

# The battle: Turn-based combat
from battle import Battle
from character import Character, Warrior, Mage

# Test play_round()
link = Warrior("Link", hp_max=120, damage=30, armor=5)
moblin = Character("Moblin", hp_max=100, damage=10)

battle = Battle(link, moblin)

print("Before one round:")
print(f"Link HP: {link.hp}")
print(f"Moblin HP: {moblin.hp}")

battle.play_round()

print("\nAfter one round:")
print(f"Link HP: {link.hp}")
print(f"Moblin HP: {moblin.hp}")

# Test simulate_battle()
link2 = Warrior("Link", hp_max=120, damage=30, armor=5)
moblin2 = Character("Moblin", hp_max=100, damage=10)

battle2 = Battle(link2, moblin2)
battle2.simulate_battle()

print("\nAfter the full battle:")
print(f"Link HP: {link2.hp}")
print(f"Moblin HP: {moblin2.hp}")
if link2.is_alive:
    print(f"{link2.name} won the battle!")
    print(f"{moblin2.name} is dead!")
else:
    print(f"{moblin2.name} won the battle!")
    print(f"{link2.name} is dead!")
```

This section demonstrates turn-based combat using the Battle class. The play_round() method lets the hero attack first, while simulate_battle() repeats the rounds until one character is defeated. The example shows a battle between Link and Moblin.
