```python
# Main working folder 
%load_ext autoreload
%autoreload 2 
from character import Character
from dice import Dice
from items import Weapon

# Presentation of each class, attributes and methods.
# Character: 
char = Character("Link", hp_max = 120, damage = 30)
print(char)
print(f"Is {char.name} alive? {char.is_alive}")
char.take_damage(50)
print(f"{char.name} took damage, new HP: {char.hp}")
char.heal(10)
print(f"{char.name} healed, new HP: {char.hp}")
```

```python
# Dice: 
d6 = Dice(6)
d20 = Dice(20)
print(f"Rolling a dice with 6 sides: {d6.roll()}")
print(f"Rolling a dice with 20 sides: {d20.roll()}")
```

```python
# Weapon:
sword = Weapon("Master Sword", damage_dice = d6)
char_w_weapon = Character("Link", hp_max = 120, damage = 30, weapon = sword) 
print(char_w_weapon)
moblin = Character("Moblin", hp_max = 56, damage = 10)
char_w_weapon.attack(moblin)
print(f"Moblin took damage after attack, new HP: {moblin.hp}")
```
